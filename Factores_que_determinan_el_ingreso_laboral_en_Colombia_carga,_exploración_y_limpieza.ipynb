## Objetivo

Identificar qué factores influyen en el ingreso laboral de los colombianos usando datos de la Gran Encuesta Integrada de Hogares (GEIH) del DANE, correspondientes al primer semestre de 2025 (enero a junio).

Los datos provienen de dos tablas de la GEIH:
- **Ocupados:** información laboral de las personas que trabajan (ingreso, sector, contrato, horas, etc.)
- **Características generales:** información demográfica (sexo, edad, nivel educativo, departamento, etc.)

Ambas tablas se cruzan usando las llaves `DIRECTORIO`, `SECUENCIA_P`, `ORDEN`, `HOGAR` y `MES`, que en conjunto identifican a cada persona encuestada en un mes específico.

### Cargar los datos en formato csv

Los datos se encuentran de forma independiente, por lo que cada mes tendrá su archivo correspondiente de características y ocupados, esto implica que se tendran 6 CSV por categoría.

```python
BASE = '/content/drive/MyDrive/Factores que determinan el ingreso laboral en Colombia'

meses = ['Enero 2025', 'Febrero 2025', 'Marzo 2025',
         'Abril 2025', 'Mayo 2025', 'Junio 2025']

ocupados = {}

for mes in meses:
    ruta = f'{BASE}/{mes}/CSV/Ocupados.CSV'
    ocupados[mes] = pd.read_csv(ruta, sep=';', encoding='latin-1', low_memory=False)
    print(f'✅ Ocupados {mes}: {len(ocupados[mes]):,} filas, {ocupados[mes].shape[1]} columnas')

print(f'\n📊 Total de meses cargados: {len(ocupados)}')
```

```python
ocupados['Febrero 2025'].head()
```

```python
CSV_CARACT = 'Características generales, seguridad social en salud y educación.CSV'

caracteristicas = {}

for mes in meses:
    ruta = f'{BASE}/{mes}/CSV/{CSV_CARACT}'
    caracteristicas[mes] = pd.read_csv(ruta, sep=';', encoding='latin-1', low_memory=False)
    print(f'✅ Características {mes}: {len(caracteristicas[mes]):,} filas, {caracteristicas[mes].shape[1]} columnas')

print(f'\n📊 Total de meses cargados: {len(caracteristicas)}')
```

```python
caracteristicas['Febrero 2025'].head()
```

### Verificación de columnas y unión de los archivos CSV

Se verifica que los archivos de cada categoría tengan las mismas columnas y en el mismo orden para luego concatenar verticalmente con el obejtivo de obtener solo una tabla por categoría.

```python
# Verificar Ocupados
columnas_ref_oc = list(ocupados['Enero 2025'].columns)
print('=== OCUPADOS ===')
for mes, df in ocupados.items():
    mismas = list(df.columns) == columnas_ref_oc
    print(f'  {mes}: {df.shape[1]} columnas — {"✅ Mismas y mismo orden" if mismas else "❌ Diferente"}')
    if not mismas:
        # Revisar si son las mismas pero en otro orden
        if set(df.columns) == set(columnas_ref_oc):
            print(f'    ⚠️ Mismas columnas pero en orden diferente')
        else:
            faltan = set(columnas_ref_oc) - set(df.columns)
            sobran = set(df.columns) - set(columnas_ref_oc)
            if faltan: print(f'    Faltan: {faltan}')
            if sobran: print(f'    Sobran: {sobran}')

# Verificar Características
columnas_ref_cg = list(caracteristicas['Enero 2025'].columns)
print('\n=== CARACTERÍSTICAS ===')
for mes, df in caracteristicas.items():
    mismas = list(df.columns) == columnas_ref_cg
    print(f'  {mes}: {df.shape[1]} columnas — {"✅ Mismas y mismo orden" if mismas else "❌ Diferente"}')
    if not mismas:
        if set(df.columns) == set(columnas_ref_cg):
            print(f'    ⚠️ Mismas columnas pero en orden diferente')
        else:
            faltan = set(columnas_ref_cg) - set(df.columns)
            sobran = set(df.columns) - set(columnas_ref_cg)
            if faltan: print(f'    Faltan: {faltan}')
            if sobran: print(f'    Sobran: {sobran}')
```

```python
orden = ['Enero 2025', 'Febrero 2025', 'Marzo 2025',
         'Abril 2025', 'Mayo 2025', 'Junio 2025']

ocupados_total = pd.concat([ocupados[mes] for mes in orden], ignore_index=True)
caracteristicas_total = pd.concat([caracteristicas[mes] for mes in orden], ignore_index=True)

print(f'Ocupados total:          {len(ocupados_total):,} filas x {ocupados_total.shape[1]} columnas')
print(f'Características total:   {len(caracteristicas_total):,} filas x {caracteristicas_total.shape[1]} columnas')

#print(f'\nRegistros por mes (Ocupados):')
#print(df_ocupados_total.groupby('MES', sort=False).size())
```

### Selección de columnas y cruce de tablas

Dado que en conjunto, ambas tablas tienen un total de 257 columnas, se seleccionaran solo las columnas necesarias de cada categoría y posteriormente se uniran las dos tablas por medio de sus llaves primarias formando una tabla unificada con la información necesaria para el análisis

```python
# Columnas que necesitamos de Ocupados
cols_ocupados = [
    'DIRECTORIO', 'SECUENCIA_P', 'ORDEN', 'HOGAR', 'MES',  # llaves
    'CLASE', 'DPTO', 'FEX_C18',                              # ubicación y factor de expansión
    'INGLABO', 'P6800',                                      # ingreso y horas
    'RAMA2D_R4',                                               # sector económico
    'P6430',                                                   # posición ocupacional
    'P6440', 'P6450', 'P6460',                                # contrato
    'P6920', 'P6990',                                          # pensión y ARL
    'P6426', 'P3069'                                           # antigüedad y tamaño empresa
]

# Columnas que necesitamos de Características
cols_caract = [
    'DIRECTORIO', 'SECUENCIA_P', 'ORDEN', 'HOGAR', 'MES',  # llaves
    'P3271',                                                  # sexo
    'P6040',                                                  # edad
    'P3042'                                                  # nivel educativo
]

ocupados_sel = ocupados_total[cols_ocupados].copy()
caract_sel = caracteristicas_total[cols_caract].copy()

print(f'Ocupados seleccionado:          {ocupados_sel.shape[0]:,} filas x {ocupados_sel.shape[1]} columnas')
print(f'Características seleccionado:   {caract_sel.shape[0]:,} filas x {caract_sel.shape[1]} columnas')
```

```python
ocupados_sel.head()
```

```python
caract_sel.head()
```

```python
llaves = ['DIRECTORIO', 'SECUENCIA_P', 'ORDEN', 'HOGAR', 'MES']

datos = ocupados_sel.merge(caract_sel, on=llaves, how='left')

print(f'Ocupados antes del cruce:   {len(ocupados_sel):,} filas')
print(f'Después del cruce:          {len(datos):,} filas')
print(f'Columnas:                   {datos.shape[1]}')

# Verificar que el cruce funcionó bien
sin_match = datos['P6040'].isna().sum()
pct_match = (1 - sin_match / len(datos)) * 100
print(f'\nPersonas con datos demográficos: {pct_match:.1f}%')
print(f'Personas sin match: {sin_match:,}')
```
